```python
import base64

class HYPER_Grader:
    def __init__(self):
        self.answers = {
            1: b"MTMw",   
            2: b"YWJj",   
            3: b"MTA=",   
            4: b"Ng==",   
            5: b"MTE="    
        }

    def check(self, task_id, user_answer):
        expected = base64.b64decode(self.answers[task_id]).decode().strip()
        result = str(user_answer).strip()

        if result == expected:
            print(f"Task {task_id}: ✅ 정답입니다!")
        else:
            print(f"Task {task_id}: ❌ 틀렸습니다.")
            print("당신의 출력:", result)
            print("정답:", expected)

grader = HYPER_Grader()
print("채점 준비 완료!")
```
